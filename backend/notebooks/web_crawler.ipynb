## Import libraries

```python
from dotenv import load_dotenv
from pathlib import Path
```

```python
from expenses_counter.utils.web_crawler import Crawler
from expenses_counter.services.daos import UserDAO
from expenses_counter.config import AppConfig
from expenses_counter.services.database import AsyncDatabaseClient
from expenses_counter.commands.crawl_url import CreateTransactionsFromCrawledDataCommand
```

## Initialize constants

```python
load_dotenv(Path("../.env"))
config = AppConfig()
```

```python
config.services.database.host = str(config.info.paths_info.root / "expenses.sqlite3")
db_client = AsyncDatabaseClient(config)
```

```python
url = ""
```

## Execute code

#### Web Crawler

```python
crawler = Crawler(url=url)
```

```python
data = crawler.run()
```

```python
print(data)
```

#### Save to the DB command

```python
user_dao = UserDAO(db_client)
```

```python
user = await user_dao.get_by_username("admin")
```

```python
cmd = CreateTransactionsFromCrawledDataCommand(data=data, user=user)
```

```python
await cmd.initialize()
await cmd.validate()
```

```python
await cmd.execute()
```
